```sql
SELECT  *  from samples.bakehouse.sales_customers;
SELECT  *  from samples.bakehouse.sales_franchises;
SELECT  *  from samples.bakehouse.sales_suppliers;
SELECT  *  from samples.bakehouse.sales_transactions;

SELECT s.franchiseID,f.name,sum(s.quantity) as sales_amount from samples.bakehouse.sales_transactions s join 
samples.bakehouse.sales_franchises f on s.franchiseID=f.franchiseID
group by s.franchiseID,f.name
order by sales_amount desc;



```
